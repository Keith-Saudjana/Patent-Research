```python
import ast
import pandas as pd
```

```python
def try_literal_eval(x):
    if isinstance(x, str):
        x = x.strip()
        if (x.startswith("[") and x.endswith("]")) or \
           (x.startswith("{") and x.endswith("}")) or \
           (x.startswith("(") and x.endswith(")")):
            try:
                return ast.literal_eval(x)
            except (ValueError, SyntaxError):
                return x
    return x
```

```python
import os
import shutil


def delete_folder(folder_path: str, missing_ok: bool = True) -> None:
    """
    Delete an entire folder and everything inside it.

    Parameters
    ----------
    folder_path : str
        Path to the folder you want to delete.

    missing_ok : bool
        If True, do nothing when the folder does not exist.
        If False, raise FileNotFoundError.
    """

    if not os.path.exists(folder_path):
        if missing_ok:
            return
        raise FileNotFoundError(f"Folder does not exist: {folder_path}")

    if not os.path.isdir(folder_path):
        raise NotADirectoryError(f"Path is not a folder: {folder_path}")

    shutil.rmtree(folder_path)

delete_folder("/home/yishin/keith/patent_research/image_crops")
```
