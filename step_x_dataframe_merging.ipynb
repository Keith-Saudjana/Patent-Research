```python
def try_literal_eval(x):
    if isinstance(x, str):
        x = x.strip()
        if (x.startswith("[") and x.endswith("]")) or \
           (x.startswith("{") and x.endswith("}")) or \
           (x.startswith("(") and x.endswith(")")):
            try:
                return ast.literal_eval(x)
            except (ValueError, SyntaxError):
                return x
    return x
```

```python
Crops_df = pd.read_csv(os.path.join(dataset_dir, "Impact_Sub_Crops.csv"), encoding="utf-8")
# Crops_df = Crops_df.map(try_literal_eval)

Crops_df.head()
```

```python
Keyword_df = Keyword_df.copy()
Crops_df = Crops_df.copy()

Keyword_df["original_order"] = range(len(Keyword_df))
```

```python
Merged_df = pd.merge(
    Keyword_df,
    Crops_df,
    on=[
        "title",
        "caption",
        "image_paths",
        "best_fig_desc",
        "Loc_class",
        "main_class",
        "sub_class",
    ],
    how="outer",
    suffixes=("_keyword", "_crops")
)
```

```python
Merged_df = Merged_df.map(try_literal_eval)

Merged_df.head()
```

```python
Merged_df.to_csv(
    os.path.join(dataset_dir, "Impact_Sub_Complete.csv"),
    index=False
)
```
